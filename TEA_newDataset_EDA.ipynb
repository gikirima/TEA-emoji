```python
import pandas as pd

df = pd.read_csv('Dataset/dataset_emoji_only.csv')
df.head()
```

# **EDA**

Distribusi kelas

```python
import matplotlib.pyplot as plt

# Select the emotion columns
emotion_cols = ['joy', 'trust', 'fear', 'surprise', 'sadness', 'disgust', 'anger', 'anticipation']

# Calculate the sum of each emotion column
emotion_counts = df[emotion_cols].sum()

# Create a bar plot
plt.figure(figsize=(15, 12), dpi=100)
emotion_counts.plot(kind='bar', width=0.7)
plt.title('Distribution of Emotions', fontsize=24)
plt.xlabel('Emotion', fontsize=20)
plt.ylabel('Count', fontsize=20)
plt.xticks(rotation=45, ha='right', fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()
```

Distribusi Sumber

```python
df['source'] = df['source'].replace('twitter.com', 'x.com')
display(df['source'].value_counts())
```

```python
# Calculate the count of each source
source_counts = df['source'].value_counts()

# Create a bar plot
plt.figure(figsize=(15, 12), dpi=100)
source_counts.plot(kind='bar', width=0.9)
plt.title('Distribution of Sources', fontsize=16)
plt.xlabel('Source', fontsize=14)
plt.ylabel('Count', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=14)
plt.yticks(fontsize=14)
plt.tight_layout()
plt.show()
```
