```python
import pandas as pd
import re

# Load dataset
data = pd.read_csv('https://raw.githubusercontent.com/gikirima/TEA-emoji/refs/heads/main(Backup-v1.0)/dataset_emoji_only.csv')

def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Replace @username
    text = re.sub(r'@[\w_]+', ' @USER ', text)
    
    # 2. Replace Link (PERBAIKAN DI SINI)
    # Penjelasan Regex:
    # (https?://|www\.)  -> Cari http://, https://, atau www.
    # [\w\-\./?=&%~+]+   -> Ikuti dengan karakter valid URL:
    #                       \w (huruf/angka), - (minus), . (titik), / (slash), 
    #                       ? (tanda tanya), = (sama dengan), & (dan), % (persen),
    #                       ~ (tilde), + (plus).
    # Emoji TIDAK ada di daftar karakter di atas, jadi regex akan berhenti saat ketemu emoji.
    
    pattern_url = r'(https?://|www\.)[\w\-\./?=&%~+]+'
    text = re.sub(pattern_url, ' LINK ', text)
    
    # 3. Rapikan spasi
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# Terapkan fungsi
data['text'] = data['text'].apply(clean_text)

# Cek sampel hasilnya sebelum save (untuk memastikan emoji masih ada)
print("Contoh data setelah dibersihkan:")
print(data['text'].head())
```

```python
# Simpan ke file baru (disarankan agar data asli tidak rusak jika salah)
data.to_csv('dataset_emoji_only_cleaned.csv', index=False)
```

```python
# cek apakah baris memiliki emoji
import emoji
def contains_emoji(s):
    return any(char in emoji.EMOJI_DATA for char in s)
data['has_emoji'] = data['text'].apply(contains_emoji)

# print jumlah baris tanpa emoji
no_emoji_count = len(data) - data['has_emoji'].sum()
print(f"Jumlah baris tanpa emoji setelah pembersihan: {no_emoji_count}")

# print baris tanpa emoji
print("Baris tanpa emoji setelah pembersihan:")
print(data[data['has_emoji'] == False]['text'])
```
