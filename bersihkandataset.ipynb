```python
# load dataset dari Dataset/dataset_emoji_only.csv
import pandas as pd
data = pd.read_csv('Dataset/dataset_emoji_only.csv')

# bersihkan kolom text dengan mereplace @username menjadi @USER dan link menjadi LINK
data['text'] = data['text'].str.replace(r'@[\w_]+', '@USER', regex=True)
data['text'] = data['text'].str.replace(r'http\S+|www.\S+', 'LINK', regex=True)

# simpan dataset yang sudah dibersihkan ke Dataset/dataset_emoji_only.csv
data.to_csv('Dataset/dataset_emoji_only.csv', index=False)
```
