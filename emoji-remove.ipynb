```python
import pandas as pd

df_train = pd.read_csv('Dataset/emoji_train_dataset.csv')
df_test = pd.read_csv('Dataset/emoji_test_dataset.csv')
df_eval = pd.read_csv('Dataset/emoji_eval_dataset.csv')

import emoji

def remove_emojis(text):
    return emoji.replace_emoji(text, replace='')

df_train['text'] = df_train['text'].apply(remove_emojis)
df_test['text'] = df_test['text'].apply(remove_emojis)
df_eval['text'] = df_eval['text'].apply(remove_emojis)

# export datasets
df_train.to_csv('Dataset/train_dataset_no_emoji.csv', index=False)
df_test.to_csv('Dataset/test_dataset_no_emoji.csv', index=False)
df_eval.to_csv('Dataset/eval_dataset_no_emoji.csv', index=False)
```
